## Search entrez

### Subtask:
Perform a search on the NCBI database (e.g., "nucleotide" or "gene") using a specific query to find the desired mRNA information.


**Reasoning**:
Perform a search on the NCBI nucleotide database using a specific query to find mRNA information, extract the IDs, and display the count and first few IDs.



```python
search_query = "Homo sapiens[Organism] AND alpha-amylase[Title] AND mrna[Title]"
handle = Entrez.esearch(db="nucleotide", term=search_query)
record = Entrez.read(handle)
handle.close()
id_list = record["IdList"]
count = record["Count"]
print(f"Found {count} records.")
print("First 10 IDs:", id_list[:10])
```

## Fetch records

### Subtask:
Fetch the detailed records for the search results.


**Reasoning**:
Use Entrez.efetch to retrieve detailed records for the IDs obtained in the previous step, and then read and store the results.



```python
handle = Entrez.efetch(db="nucleotide", id=id_list, rettype="gb", retmode="text")
mrna_records = handle.read()
handle.close()
print(mrna_records[:500]) # Print the first 500 characters to check the content
```

## Parse records

### Subtask:
Parse the fetched records to extract the relevant mRNA information.


**Reasoning**:
Parse the fetched GenBank records and extract the relevant information into a list of dictionaries.



```python
from Bio import SeqIO
from io import StringIO

mrna_info = []
for record in SeqIO.parse(StringIO(mrna_records), "genbank"):
    mrna_info.append({
        "accession": record.id,
        "description": record.description,
        "sequence": str(record.seq)
    })

print(f"Extracted information for {len(mrna_info)} records.")
if mrna_info:
    print("First record:", mrna_info[0])
```

## Display information

### Subtask:
Display the retrieved mRNA information.


**Reasoning**:
Iterate through the mrna_info list and print the requested information for each record.



```python
for record in mrna_info:
    print("Accession:", record["accession"])
    print("Description:", record["description"])
    print("Sequence (first 100 chars):", record["sequence"][:100])
    print("-" * 20)
```
